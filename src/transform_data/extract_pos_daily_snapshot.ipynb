```python
import sys

import polars as pl
from loguru import logger
```

```python
logger.remove()
logger.add(sys.stderr, level='INFO')
```

# Read data

```python
path = "data/processed/pos.xlsx"

pos = pl.read_excel(path)

pos.head()
```

# Process

```python
pos_daily_snapshot = (
    pos
    .with_columns(
        pl.col('datetime').dt.date().alias('date')
    )
    .group_by('restaurant', 'date')
    .agg(
        pl.col('pcs').sum(),
        pl.concat_list('src').flatten().unique()
    )
    .with_row_index('id')
)

pos_daily_snapshot.head()
```

# Save

```python
path = "data/processed/pos_daily_snapshot.parquet"

pos_daily_snapshot.write_parquet(path)
```
